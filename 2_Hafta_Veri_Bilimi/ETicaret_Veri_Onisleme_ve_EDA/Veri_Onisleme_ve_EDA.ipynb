```python
import pandas as pd

# Veri setini okuma
df = pd.read_csv('e_ticaret_veri_seti - Kopya.csv')

# Soru 1: İlk 10 satır ve son 10 satırı yazdırma
print(" - İLK 10 SATIR -")
display(df.head(10))

print("\n - SON 10 SATIR - ")
display(df.tail(10))

# Soru 2: Satır sayısı, sütun sayısı ve sütun isimleri
print(f"\n Satır Sayısı: {df.shape[0]}")
print(f" Sütun Sayısı: {df.shape[1]}")
print("\n Sütun İsimleri:")
print(df.columns.tolist())
```

```python
# Soru 3: info() ve describe() çıktıları, sayısal/kategorik sütun ayrımı

print("  Veri Seti Bilgisi (info)  ")
df.info()

print("\n  İstatistiksel Özet (describe)  ")
display(df.describe())

# Sayısal ve kategorik sütunları filtreleme
sayisal_sutunlar = df.select_dtypes(include=['number']).columns.tolist()
kategorik_sutunlar = df.select_dtypes(include=['object', 'category']).columns.tolist()

print(f"\n Sayısal Sütunlar: {sayisal_sutunlar}")
print(f" Kategorik Sütunlar: {kategorik_sutunlar}")
```

```python
# Soru 4: Eksik değer sayısı, oranı ve en yüksek 5 sütun

eksik_sayilari = df.isnull().sum()
eksik_oranlari = (df.isnull().sum() / len(df)) * 100

# Tablo haline getirme
eksik_df = pd.DataFrame({
    'Eksik Sayısı': eksik_sayilari,
    'Eksik Oranı (%)': eksik_oranlari
})

print(" Eksik Veri Oranı En Yüksek 5 Sütun ")
display(eksik_df.sort_values(by='Eksik Oranı (%)', ascending=False).head(5))
```

```python
# Soru 5: En az bir eksik değer içeren satırları filtreleme ve ilk 20'sini gösterme

# axis=1 ile satır bazında en az bir tane NaN içerenleri buluyoruz
eksik_iceren_satirlar = df[df.isnull().any(axis=1)]

print(f"En az bir eksik değer içeren toplam satır sayısı: {len(eksik_iceren_satirlar)}")
print("\n  Eksik Değer İçeren İlk 20 Satır ")
display(eksik_iceren_satirlar.head(20))
```

```python
# Soru 6: indirim_orani ve musteri_puani sütunlarındaki eksik değerleri medyan ile doldurma

sutunlar_medyan = ['indirim_orani', 'musteri_puani']

print(" Doldurmadan Önce Eksik Değer Sayıları ")
display(df[sutunlar_medyan].isnull().sum())

# Her iki sütunu kendi medyan değeri ile dolduruyoruz
for sutun in sutunlar_medyan:
    medyan_degeri = df[sutun].median()
    df[sutun] = df[sutun].fillna(medyan_degeri)

print("\n  Doldurduktan Sonra Eksik Değer Sayıları ")
display(df[sutunlar_medyan].isnull().sum())
```

```python
# Soru 7: odeme_turu ve musteri_tipi sütunlarındaki eksik değerleri mod ile doldurma

sutunlar_mod = ['odeme_turu', 'musteri_tipi']

print("   Doldurmadan Önce Eksik Değer Sayıları ")
display(df[sutunlar_mod].isnull().sum())

# Her iki kategorik sütunu en çok tekrar eden değer (mod) ile dolduruyoruz
for sutun in sutunlar_mod:
    mod_degeri = df[sutun].mode()[0]  # mode() fonksiyonu liste döner, [0] ile ilkini (en sık olanı) alıyoruz
    df[sutun] = df[sutun].fillna(mod_degeri)

print("\n   Doldurduktan Sonra Eksik Değer Sayıları (Kontrol) ")
display(df[sutunlar_mod].isnull().sum())
```

```python
# Soru 8: siparis_tarihi sütununu datetime tipine dönüştürme ve yeni özellikler üretme

# Sütunu datetime formatına dönüştürüyoruz
df['siparis_tarihi'] = pd.to_datetime(df['siparis_tarihi'])

# Yıl, ay, gün ve haftanın günü sütunlarını üretiyoruz
df['siparis_yili'] = df['siparis_tarihi'].dt.year
df['siparis_ayi'] = df['siparis_tarihi'].dt.month
df['siparis_gunu'] = df['siparis_tarihi'].dt.day
df['haftanin_gunu'] = df['siparis_tarihi'].dt.day_name()

print(" Yeni Üretilen Tarih Sütunlarının İlk 10 Satırı ")
display(df[['siparis_tarihi', 'siparis_yili', 'siparis_ayi', 'siparis_gunu', 'haftanin_gunu']].head(10))
```

```python
# Soru 9: Duplicate (tekrarlayan) kayıtları bulma ve temizleme

tekrar_eden_sayisi = df.duplicated().sum()
print(f"Temizlik Öncesi Veri Seti Satır Sayısı: {len(df)}")
print(f"Bulunan Tekrarlayan (Duplicate) Kayıt Sayısı: {tekrar_eden_sayisi}")

# Tekrarlayan kayıtları veri setinden tamamen siliyoruz
df = df.drop_duplicates()

print(f"Temizlik Sonrası Veri Seti Satır Sayısı: {len(df)}")
```

```python
# Soru 10: sehir sütunundaki benzersiz değerleri inceleme ve standartlaştırma

print(f" Temizlik Öncesi Eşsiz Şehir Sayısı: {df['sehir'].nunique()} ")
print(df['sehir'].unique())

# Metinleri standart hale getirme: Başındaki/sonundaki boşlukları sil (strip) ve baş harfleri büyük yap (title)
df['sehir'] = df['sehir'].str.strip().str.title()

# İngilizce "I" harfi ile yazılan Istanbul'u Türkçe karaktere çevirme gibi manuel düzeltmeler
df['sehir'] = df['sehir'].replace({'Istanbul': 'İstanbul'})

print(f"\n Temizlik Sonrası Eşsiz Şehir Sayısı: {df['sehir'].nunique()} ")
print(df['sehir'].unique())
```

```python
# Soru 11: kategori sütunundaki tutarsız değerleri tek biçime dönüştürme

print(" Temizlik Öncesi Kategori Dağılımı ")
display(df['kategori'].value_counts())

# Boşlukları sil, küçük harfe çevir, '&' işaretini boşluk yap, çift boşluğu teke indir ve baş harfleri büyüt
df['kategori'] = df['kategori'].str.strip().str.lower().str.replace('&', ' ').str.replace('  ', ' ').str.title()

# Manuel düzeltmeler (Türkçe karakter hatası ve bağlaç fazlalığı)
df['kategori'] = df['kategori'].replace({
    'Ev Ve Yaşam': 'Ev Yaşam',
    'Kozmetık': 'Kozmetik'
})

print("\n Temizlik Sonrası Kategori Dağılımı ")
display(df['kategori'].value_counts())
```

```python
# Soru 12: Mantıksal hatalı kayıtları bulma ve çıkarma

# Hatalı koşulları birleştirerek filtreleme yapıyoruz
hatali_kayitlar = df[(df['birim_fiyat'] <= 0) |
                     (df['toplam_tutar'] <= 0) |
                     (df['teslimat_gunu'] < 0) |
                     (df['musteri_puani'] > 5)]

print(f"Mantıksal olarak hatalı kayıt sayısı: {len(hatali_kayitlar)}")

# Hatalı DEĞİL (~) koşulu ile temiz veriyi df üzerine yazıyoruz
df = df[~((df['birim_fiyat'] <= 0) |
          (df['toplam_tutar'] <= 0) |
          (df['teslimat_gunu'] < 0) |
          (df['musteri_puani'] > 5))]

print(f"Hatalı kayıtlar silindikten sonra kalan satır sayısı: {len(df)}")
```

```python
# Soru 13: birim_fiyat sütunu için IQR hesaplama ve aykırı değerleri (outlier) bulma

Q1 = df['birim_fiyat'].quantile(0.25)
Q3 = df['birim_fiyat'].quantile(0.75)
IQR = Q3 - Q1

alt_sinir = Q1 - 1.5 * IQR
ust_sinir = Q3 + 1.5 * IQR

print(f"Q1 (1. Çeyreklik): {Q1}")
print(f"Q3 (3. Çeyreklik): {Q3}")
print(f"IQR: {IQR}")
print(f"Alt Sınır: {alt_sinir}")
print(f"Üst Sınır: {ust_sinir}")

# Sınırların dışında kalan aykırı değerleri filtreleme
aykiri_degerler = df[(df['birim_fiyat'] < alt_sinir) | (df['birim_fiyat'] > ust_sinir)]
print(f"\nAykırı değer içeren kayıt sayısı: {len(aykiri_degerler)}")
```

```python
# Gerekli görselleştirme kütüphanelerini içe aktarma
import matplotlib.pyplot as plt
import seaborn as sns

# Soru 14: toplam_tutar sütunu için histogram ve box plot çizdirme

plt.figure(figsize=(14, 4))

# Histogram
plt.subplot(1, 2, 1)
sns.histplot(df['toplam_tutar'], kde=True, bins=30, color='skyblue')
plt.title('Toplam Tutar - Histogram Dağılımı')
plt.xlabel('Toplam Tutar')
plt.ylabel('Frekans')

# Box Plot (Kutu Grafiği)
plt.subplot(1, 2, 2)
sns.boxplot(x=df['toplam_tutar'], color='lightgreen')
plt.title('Toplam Tutar - Box Plot (Aykırı Değerler)')
plt.xlabel('Toplam Tutar')

plt.tight_layout()
plt.show()
```

```python
# Soru 15: Kategorik/Sayısal analizler ve iş sorularının cevaplanması

# Müşteri Tipi Temizliği
df['musteri_tipi'] = df['musteri_tipi'].str.strip().str.title()
df['musteri_tipi'] = df['musteri_tipi'].replace({'Vip': 'VIP'})

# Ödeme Türü Temizliği
df['odeme_turu'] = df['odeme_turu'].str.strip().str.title()

odeme_sozlugu = {
    'Kredi Karti': 'Kredi Kartı',
    'Banka Karti': 'Banka Kartı',
    'Kapida Odeme': 'Kapıda Ödeme',
    'Kapida Ödeme': 'Kapıda Ödeme',
    'Kapıda Odeme': 'Kapıda Ödeme',
    'Havale/Eft': 'Havale/EFT',
    'Havale': 'Havale/EFT',
    'Eft': 'Havale/EFT',
    'Kart': 'Kredi Kartı'
}
df['odeme_turu'] = df['odeme_turu'].replace(odeme_sozlugu)

# Tabloları Yazdırma
print("Kategori Sütunu value_counts() Çıktısı:")
display(df['kategori'].value_counts())

print("\n Müşteri Tipi Yüzde Dağılımı: ")
display(df['musteri_tipi'].value_counts(normalize=True) * 100)

print("\n Temel İstatistikler (birim_fiyat, toplam_tutar, musteri_puani): ")
display(df[['birim_fiyat', 'toplam_tutar', 'musteri_puani']].describe())

# Grafik Çizimi
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='odeme_turu', hue='odeme_turu', palette='viridis', legend=False)
plt.title('Ödeme Türlerine Göre Sipariş Sayıları')
plt.xlabel('Ödeme Türü')
plt.ylabel('Sipariş Sayısı')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# İş Sorularının Cevapları
print("\n İş Sorularının Cevapları:")
en_yuksek_kategori = df.groupby('kategori')['toplam_tutar'].mean().idxmax()
print(f"1) En yüksek ortalama toplam tutara sahip kategori: {en_yuksek_kategori}")

en_yuksek_musteri = df.groupby('musteri_tipi')['musteri_puani'].mean().idxmax()
print(f"2) En yüksek ortalama müşteri puanına sahip müşteri tipi: {en_yuksek_musteri}")
```
